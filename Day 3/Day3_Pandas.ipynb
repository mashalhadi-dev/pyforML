```python
import pandas as pd
import numpy as np

# 1. Aisa data banate hain jisme kuch values gayab (NaN) hon
messy_data = {
    "Product": ["Laptop", "Mouse", "Keyboard", "Monitor", "Printer"],
    "Price": [1200, 25, np.nan, 300, np.nan],  # Yahan 2 prices gayab hain (NaN)
    "Stock": [10, 150, 80, np.nan, 20]          # Yahan aik stock gayab hai
}

df_messy = pd.DataFrame(messy_data)
print("--- Khali data (Original Messy DataFrame) ---")
display(df_messy)

# 2. Yeh check karna ke kahan kahan data missing hai
print("\n--- Kahan kahan values missing hain? (isnull) ---")
display(df_messy.isnull())

# 3. Method A: dropna() - Jidhar bhi NaN ho, us poori row ko ura do
df_dropped = df_messy.dropna()
print("\n--- dropna() ke baad (Khali rows delete ho gayeen) ---")
display(df_dropped)

# 4. Method B: fillna() - Khali jagah par intelligent value dalna (Misal ke tor par Average Price)
mean_price = df_messy["Price"].mean()  # Price ka average nikala
df_filled = df_messy.copy()
df_filled["Price"] = df_filled["Price"].fillna(mean_price) # Khali price mein average daal diya
df_filled["Stock"] = df_filled["Stock"].fillna(0)          # Khali stock mein 0 daal diya

print("\n--- fillna() ke baad (Intelligently fix kiya gaya DataFrame) ---")
display(df_filled)
```

```python
import pandas as pd
import numpy as np

# 1. Asli Dataset ko direct URL se load karna (pd.read_csv ka kamaal)
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df_titanic = pd.read_csv(url)

print("--- Titanic Dataset Successfully Load Ho Gaya! ---")
# Sirf pehli 5 rows dekhne ke liye .head() use karte hain
display(df_titanic.head())

# 2. Memory Profiling: X-ray report check karna ke RAM kitni use ho rahi hai
print("\n--- Memory & Data Info (df.info) ---")
df_titanic.info()

# 3. Yeh check karna ke kis column mein kitni values missing hain
print("\n--- Missing Values Count per Column ---")
print(df_titanic.isnull().sum())

# Ab hum missing values ko intelligent tareeqay se fix karenge

# 1. Age ke column mein jo values missing hain, wahan sab ki median (beech wali) age daal dete hain
median_age = df_titanic["Age"].median()
df_titanic["Age"] = df_titanic["Age"].fillna(median_age)

# 2. Embarked column mein jo 2 values missing hain, wahan sab se zyada aane wala shehar (mode) daal dete hain
most_common_embarked = df_titanic["Embarked"].mode()[0]
df_titanic["Embarked"] = df_titanic["Embarked"].fillna(most_common_embarked)

# 3. Cabin column mein bohot zyada values missing hain, toh wahan 'Unknown' likh kar dabba bhar dete hain
df_titanic["Cabin"] = df_titanic["Cabin"].fillna("Unknown")

# Dobara check karte hain ke ab koi missing value bachi hai ya nahi!
print("--- Cleaning ke baad Missing Values Count ---")
print(df_titanic.isnull().sum())
```

```python
# 1. Dekhte hain ke mardon aur aurton mein se kitne feesad log bache (Survived)
survival_by_sex = df_titanic.groupby('Sex')['Survived'].mean() * 100
print("--- Gender ke lihaz se Survival Percentage ---")
print(survival_by_sex)

# 2. Yeh dekhte hain ke Ticket Class ke lihaz se kya haal tha (1st class vs 3rd class)
survival_by_class = df_titanic.groupby('Pclass')['Survived'].mean() * 100
print("\n--- Ticket Class ke lihaz se Survival Percentage ---")
print(survival_by_class)
```
