```python
import pandas as pd
import numpy as np

# 1. Aisa data banate hain jisme kuch values gayab (NaN) hon
messy_data = {
    "Product": ["Laptop", "Mouse", "Keyboard", "Monitor", "Printer"],
    "Price": [1200, 25, np.nan, 300, np.nan],  # Yahan 2 prices gayab hain (NaN)
    "Stock": [10, 150, 80, np.nan, 20]          # Yahan aik stock gayab hai
}

df_messy = pd.DataFrame(messy_data)
print("--- Khali data (Original Messy DataFrame) ---")
display(df_messy)

# 2. Yeh check karna ke kahan kahan data missing hai
print("\n--- Kahan kahan values missing hain? (isnull) ---")
display(df_messy.isnull())

# 3. Method A: dropna() - Jidhar bhi NaN ho, us poori row ko ura do
df_dropped = df_messy.dropna()
print("\n--- dropna() ke baad (Khali rows delete ho gayeen) ---")
display(df_dropped)

# 4. Method B: fillna() - Khali jagah par intelligent value dalna (Misal ke tor par Average Price)
mean_price = df_messy["Price"].mean()  # Price ka average nikala
df_filled = df_messy.copy()
df_filled["Price"] = df_filled["Price"].fillna(mean_price) # Khali price mein average daal diya
df_filled["Stock"] = df_filled["Stock"].fillna(0)          # Khali stock mein 0 daal diya

print("\n--- fillna() ke baad (Intelligently fix kiya gaya DataFrame) ---")
display(df_filled)
```
