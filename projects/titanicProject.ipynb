```python
import pandas as pd
import numpy as np

# 1. Train aur Test data load karna (Kaggle par do files milti hain: train.csv aur test.csv)
train_url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df_train = pd.read_csv(train_url)

print("Data successfully load ho gaya! Rows aur Columns:", df_train.shape)

# 2. Missing values ko fill karna
# Age mein median daal diya
df_train["Age"] = df_train["Age"].fillna(df_train["Age"].median())

# Embarked mein most common (mode) shehar daal diya
df_train["Embarked"] = df_train["Embarked"].fillna(df_train["Embarked"].mode()[0])

# Fare (kiraya) mein median daal diya (agar koi khali ho)
df_train["Fare"] = df_train["Fare"].fillna(df_train["Fare"].median())

print("Level 1 Complete: Data saaf suthra ho gaya hai!")
```

```python
# 1. Sex column ko numbers mein badalna (male = 0, female = 1)
df_train["Sex"] = df_train["Sex"].map({"male": 0, "female": 1})

# 2. Embarked column ko bhi numbers mein badalna (S=0, C=1, Q=2)
df_train["Embarked"] = df_train["Embarked"].map({"S": 0, "C": 1, "Q": 2})

# 3. Naya Feature banana: Family Size (SibSp + Parch + Ap khud)
df_train["FamilySize"] = df_train["SibSp"] + df_train["Parch"] + 1

print("Level 2 Complete: Text columns numbers mein badal gaye aur naye features ban gaye!")
display(df_train[["Sex", "Embarked", "FamilySize"]].head())
```

```python
from sklearn.ensemble import RandomForestClassifier

# 1. Wo columns chunna jo model ko sikhane ke liye use honge (Features)
features = ["Pclass", "Sex", "Age", "SibSp", "Parch", "Fare", "Embarked", "FamilySize"]

X = df_train[features]  # Input data
y = df_train["Survived"] # Jawab (Target: Bach gaya ya nahi)

# 2. Model ko train karna
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X, y)
print("Level 3 Complete: Model ne data parh kar training mukammal kar li hai!")

# 3. Predictions karna aur submission file banana
# (Asal Kaggle test set par prediction karne ke baad yeh code submission.csv bana dega)
predictions = model.predict(X)

# Submission DataFrame banana
submission = pd.DataFrame({
    "PassengerId": df_train["PassengerId"],
    "Survived": predictions
})

# CSV file ke taur par save karna
submission.to_csv("my_submission.csv", index=False)
print("Zabardast! 'my_submission.csv' file successfully save ho gayi hai!")
```

```python
import pandas as pd
import numpy as np

# 1. Asli Train aur Test files load karna (Kaggle ke official links se)
train_url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
train_df = pd.read_csv(train_url)

# Kaggle ka official test dataset link
test_url = "https://raw.githubusercontent.com/agconti/kaggle-titanic/master/data/test.csv"
test_df = pd.read_csv(test_url)

print("Test dataset successfully load ho gaya! Rows:", len(test_df))

# 2. Test data par bhi wahi cleaning apply karna jo train par ki thi
# Age mein median bharna
median_age = train_df["Age"].median()  # Train ka median use karein ge
test_df["Age"] = test_df["Age"].fillna(median_age)

# Fare mein median bharna (test mein aik Fare khali ho sakta hai)
median_fare = train_df["Fare"].median()
test_df["Fare"] = test_df["Fare"].fillna(median_fare)

# Cabin mein 'Unknown'
test_df["Cabin"] = test_df["Cabin"].fillna("Unknown")

# 3. Simple Prediction Rule (Misal ke tor par: Jo aurtein hain wo bachi, mard nahi)
# Hum rule banate hain ke agar Sex == 'female' toh Survived = 1, warna 0
test_df['Survived'] = 0
test_df.loc[test_df['Sex'] == 'female', 'Survived'] = 1

# 4. Sirf zaroori columns rakhna (PassengerId aur Survived)
submission = test_df[['PassengerId', 'Survived']]

# 5. Nayi submission file save karna
submission.to_csv('my_submission.csv', index=False)
print("Nayi 'my_submission.csv' file kamiyabi se ban gayi hai jisme 418 rows hain!")
```
